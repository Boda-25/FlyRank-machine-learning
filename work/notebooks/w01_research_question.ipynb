## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python

import pandas as pd

url = "https://raw.githubusercontent.com/Boda-25/FlyRank-machine-learning/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
print("Number of unique clients:", df["client_id"].nunique())

declining = df["trend_direction"].astype(str).str.lower().eq("down").sum()

print("Pages with declining trends:", declining)
print(f"Declining trends percentage: {declining / len(df) * 100:.2f}%")
```
